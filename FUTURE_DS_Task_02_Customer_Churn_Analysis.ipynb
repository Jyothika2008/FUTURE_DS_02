```python
# ============================================================
# PROFESSIONAL CUSTOMER CHURN DASHBOARD - FINAL VERSION
# ============================================================

import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

print("Building Professional Customer Churn Dashboard...")


# ============================================================
# 1. KPI CALCULATIONS
# ============================================================

total_customers = len(df)

churned_customers = (df["Churn"] == "Yes").sum()
retained_customers = (df["Churn"] == "No").sum()

churn_rate = churned_customers / total_customers * 100
retention_rate = retained_customers / total_customers * 100

avg_tenure = df["tenure"].mean()


# ============================================================
# 2. CHURN CALCULATIONS
# ============================================================

# Contract
contract_churn = (
    pd.crosstab(
        df["Contract"],
        df["Churn"],
        normalize="index"
    ) * 100
)

# Tenure
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 60, 72],
    labels=[
        "0–12 Months",
        "13–24 Months",
        "25–48 Months",
        "49–60 Months",
        "61–72 Months"
    ]
)

tenure_churn = (
    pd.crosstab(
        df["TenureGroup"],
        df["Churn"],
        normalize="index"
    ) * 100
)

# Payment method
payment_churn = (
    pd.crosstab(
        df["PaymentMethod"],
        df["Churn"],
        normalize="index"
    ) * 100
)

# Internet service
internet_churn = (
    pd.crosstab(
        df["InternetService"],
        df["Churn"],
        normalize="index"
    ) * 100
)

# Monthly charges
monthly_charges = df.groupby("Churn")["MonthlyCharges"].mean()


# ============================================================
# 3. CREATE DASHBOARD LAYOUT
# ============================================================

fig = make_subplots(

    rows=4,
    cols=3,

    row_heights=[
        0.11,   # KPI row 1
        0.11,   # KPI row 2
        0.39,   # Charts row 1
        0.39    # Charts row 2
    ],

    vertical_spacing=0.055,

    horizontal_spacing=0.055,

    specs=[

        # KPI ROW 1
        [
            {"type": "indicator"},
            {"type": "indicator"},
            {"type": "indicator"}
        ],

        # KPI ROW 2
        [
            {"type": "indicator"},
            {"type": "indicator"},
            {"type": "indicator"}
        ],

        # CHART ROW 1
        [
            {"type": "domain"},
            {"type": "xy"},
            {"type": "xy"}
        ],

        # CHART ROW 2
        [
            {"type": "xy"},
            {"type": "xy"},
            {"type": "xy"}
        ]
    ],

    subplot_titles=[
        "",
        "",
        "",
        "",
        "",
        "",
        "Customer Retention vs Churn",
        "Churn Rate by Contract",
        "Churn Rate by Tenure",
        "Churn by Payment Method",
        "Churn by Internet Service",
        "Average Monthly Charges by Churn"
    ]
)


# ============================================================
# 4. KPI CARDS
# ============================================================

fig.add_trace(

    go.Indicator(

        mode="number",

        value=total_customers,

        title={
            "text": "<b>TOTAL CUSTOMERS</b>"
        },

        number={
            "valueformat": ",",
            "font": {
                "size": 32,
                "color": "#087EA4"
            }
        },

        domain={
            "x": [0, 1],
            "y": [0, 1]
        }

    ),

    row=1,
    col=1
)


fig.add_trace(

    go.Indicator(

        mode="number",

        value=churned_customers,

        title={
            "text": "<b>CHURNED CUSTOMERS</b>"
        },

        number={
            "valueformat": ",",
            "font": {
                "size": 32,
                "color": "#E76F51"
            }
        }

    ),

    row=1,
    col=2
)


fig.add_trace(

    go.Indicator(

        mode="number",

        value=churn_rate,

        title={
            "text": "<b>CHURN RATE</b>"
        },

        number={
            "suffix": "%",
            "valueformat": ".2f",
            "font": {
                "size": 32,
                "color": "#E76F51"
            }
        }

    ),

    row=1,
    col=3
)


# ============================================================
# 5. SECOND KPI ROW
# ============================================================

fig.add_trace(

    go.Indicator(

        mode="number",

        value=retained_customers,

        title={
            "text": "<b>RETAINED CUSTOMERS</b>"
        },

        number={
            "valueformat": ",",
            "font": {
                "size": 32,
                "color": "#2A9D8F"
            }
        }

    ),

    row=2,
    col=1
)


fig.add_trace(

    go.Indicator(

        mode="number",

        value=retention_rate,

        title={
            "text": "<b>RETENTION RATE</b>"
        },

        number={
            "suffix": "%",
            "valueformat": ".2f",
            "font": {
                "size": 32,
                "color": "#2A9D8F"
            }
        }

    ),

    row=2,
    col=2
)


fig.add_trace(

    go.Indicator(

        mode="number",

        value=avg_tenure,

        title={
            "text": "<b>AVERAGE TENURE</b>"
        },

        number={
            "suffix": " months",
            "valueformat": ".1f",
            "font": {
                "size": 30,
                "color": "#087EA4"
            }
        }

    ),

    row=2,
    col=3
)


# ============================================================
# 6. RETENTION VS CHURN DONUT
# ============================================================

fig.add_trace(

    go.Pie(

        labels=[
            "Retained",
            "Churned"
        ],

        values=[
            retained_customers,
            churned_customers
        ],

        hole=0.58,

        textinfo="label+percent",

        textfont={
            "size": 11
        },

        marker={
            "colors": [
                "#2A9D8F",
                "#E76F51"
            ]
        },

        showlegend=False

    ),

    row=3,
    col=1
)


# ============================================================
# 7. CONTRACT CHURN
# ============================================================

fig.add_trace(

    go.Bar(

        x=contract_churn.index,

        y=contract_churn["Yes"],

        text=[
            f"{x:.1f}%"
            for x in contract_churn["Yes"]
        ],

        textposition="outside",

        marker_color="#3A6EA5",

        hovertemplate=
        "<b>%{x}</b><br>"
        "Churn Rate: %{y:.2f}%"
        "<extra></extra>"

    ),

    row=3,
    col=2
)


# ============================================================
# 8. TENURE CHURN
# ============================================================

fig.add_trace(

    go.Bar(

        x=tenure_churn.index,

        y=tenure_churn["Yes"],

        text=[
            f"{x:.1f}%"
            for x in tenure_churn["Yes"]
        ],

        textposition="outside",

        marker_color="#5B4B8A",

        hovertemplate=
        "<b>%{x}</b><br>"
        "Churn Rate: %{y:.2f}%"
        "<extra></extra>"

    ),

    row=3,
    col=3
)


# ============================================================
# 9. PAYMENT METHOD CHURN
# ============================================================

fig.add_trace(

    go.Bar(

        x=payment_churn["Yes"],

        y=payment_churn.index,

        orientation="h",

        text=[
            f"{x:.1f}%"
            for x in payment_churn["Yes"]
        ],

        textposition="outside",

        marker_color="#4C956C",

        hovertemplate=
        "<b>%{y}</b><br>"
        "Churn Rate: %{x:.2f}%"
        "<extra></extra>"

    ),

    row=4,
    col=1
)


# ============================================================
# 10. INTERNET SERVICE CHURN
# ============================================================

fig.add_trace(

    go.Bar(

        x=internet_churn.index,

        y=internet_churn["Yes"],

        text=[
            f"{x:.1f}%"
            for x in internet_churn["Yes"]
        ],

        textposition="outside",

        marker_color="#F4A261",

        hovertemplate=
        "<b>%{x}</b><br>"
        "Churn Rate: %{y:.2f}%"
        "<extra></extra>"

    ),

    row=4,
    col=2
)


# ============================================================
# 11. MONTHLY CHARGES BY CHURN
# ============================================================

fig.add_trace(

    go.Bar(

        x=[
            "Retained",
            "Churned"
        ],

        y=[
            monthly_charges["No"],
            monthly_charges["Yes"]
        ],

        text=[
            f"{monthly_charges['No']:.2f}",
            f"{monthly_charges['Yes']:.2f}"
        ],

        textposition="outside",

        marker_color=[
            "#2A9D8F",
            "#E76F51"
        ],

        hovertemplate=
        "<b>%{x}</b><br>"
        "Average Monthly Charge: %{y:.2f}"
        "<extra></extra>"

    ),

    row=4,
    col=3
)


# ============================================================
# 12. AXIS LABELS
# ============================================================

fig.update_yaxes(
    title_text="Churn Rate (%)",
    range=[0, 50],
    row=3,
    col=2
)

fig.update_yaxes(
    title_text="Churn Rate (%)",
    range=[0, 55],
    row=3,
    col=3
)

fig.update_xaxes(
    title_text="Churn Rate (%)",
    row=4,
    col=1
)

fig.update_yaxes(
    title_text="Payment Method",
    row=4,
    col=1
)

fig.update_yaxes(
    title_text="Churn Rate (%)",
    range=[0, 50],
    row=4,
    col=2
)

fig.update_yaxes(
    title_text="Average Charge",
    row=4,
    col=3
)


# ============================================================
# 13. PROFESSIONAL STYLING
# ============================================================

fig.update_layout(

    title={

        "text":
        "<b>CUSTOMER CHURN & RETENTION ANALYSIS</b>"
        "<br>"
        "<sup>Telecommunication Customer Analysis | Future Interns - Task 2</sup>",

        "x": 0.5,

        "xanchor": "center",

        "font": {
            "size": 22,
            "color": "#17365D"
        }
    },

    width=1250,

    height=1150,

    paper_bgcolor="#F7F9FC",

    plot_bgcolor="white",

    font={
        "family": "Arial",
        "size": 10,
        "color": "#263238"
    },

    margin={
        "l": 80,
        "r": 80,
        "t": 130,
        "b": 70
    },

    showlegend=False
)


# ============================================================
# 14. DISPLAY
# ============================================================

fig.show()
```

## 7. Key Findings
### Overall Churn and Retention
- The overall customer churn rate is **26.54%**.
- The overall customer retention rate is **73.46%**.

### Contract Type
- Month-to-month customers have the highest observed churn rate at **42.71%**.
- One-year contract customers have a churn rate of **11.27%**.
- Two-year contract customers have a churn rate of **2.83%**.

### Customer Lifetime
- Customers with **0–12 months** of tenure have the highest observed churn rate at **47.44%**.
- Churn decreases as customer tenure increases.
- Customers with **49–72 months** of tenure have a churn rate of **9.51%**.

### Monthly Charges
- Customers who churned had an average monthly charge of **74.44**.
- Customers who stayed had an average monthly charge of **61.27**.

### Internet Service
- Fiber optic customers have the highest observed churn rate at **41.89%**.
- DSL customers have a churn rate of **18.96%**.
- Customers without internet service have a churn rate of **7.40%**.

### Payment Method
- Customers using electronic check have the highest observed churn rate at **45.29%**.
- Customers using automatic credit-card payments have a churn rate of **15.24%**.

### Customer Support and Demographics
- Customers without tech support have a churn rate of **41.64%**, compared with **15.17%** among customers with tech support.
- Senior citizens have a higher observed churn rate (**41.68%**) than non-senior customers (**23.61%**).
- Gender shows very little difference in churn rate.

## 8. Recommendations
### Recommended Retention Strategies

1. **Focus on early-tenure customers**
   - Customers in their first 12 months show the highest observed churn rate.
   - Introduce onboarding support, regular follow-ups, and early engagement programs.

2. **Strengthen retention of month-to-month customers**
   - Month-to-month customers have a substantially higher observed churn rate than customers on longer contracts.
   - Consider retention offers and incentives that encourage longer-term contracts.

3. **Investigate high-churn payment methods**
   - Electronic check users show a high observed churn rate.
   - Investigate whether payment experience, convenience, or other customer characteristics are associated with this pattern.

4. **Improve support for customers without tech support**
   - Customers without tech support show a higher observed churn rate.
   - Consider targeted support options and proactive technical assistance.

5. **Investigate higher-charge customer segments**
   - Customers who churned have higher average monthly charges than customers who stayed.
   - Examine whether pricing, service combinations, or perceived value are related to this pattern.

6. **Investigate fiber-optic customer churn**
   - Fiber-optic customers show a higher observed churn rate.
   - Examine service quality, pricing, support, and customer experience within this group.

## 9. Conclusion
The customer churn analysis shows an overall churn rate of **26.54%** and a retention rate of **73.46%**. The analysis identified several customer groups with higher observed churn rates, particularly month-to-month customers, customers with shorter tenure, fiber-optic customers, and electronic-check users. Customers without tech support also showed higher churn. These findings can help the company focus its retention efforts on high-risk customer segments. Further analysis would be useful to understand the underlying reasons for these patterns and to design targeted retention strategies.
